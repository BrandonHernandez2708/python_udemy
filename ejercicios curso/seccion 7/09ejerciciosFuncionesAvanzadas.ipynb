Crea una funcion estandar para calcular el cuadrado de una lista de numeros.

```python
def cuadrado(lista,cuadrados=[]):
  for elementos in lista:
    elementos = int(elementos)
    cuadrado=pow(elementos,2)
    cuadrados.append(cuadrado)
  return cuadrados

print("Ingrese una lista para calcular sus cuadrados")
lista=[2,4,6]
print(cuadrado(lista))

```
