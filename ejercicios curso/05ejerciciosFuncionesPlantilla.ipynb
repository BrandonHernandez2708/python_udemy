Funcion que solicite al usuario nombre y edad para mostrarlos despues.

```python
def mostrar_datos():
  nombre = input("Ingrese su nombre: ")
  edad=int(input("ingrese su edad: "))
  return f" {nombre} {edad}"
print(mostrar_datos())
```
